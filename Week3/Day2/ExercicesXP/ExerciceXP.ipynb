```python
import numpy as np
# 1. Création d'une matrice 3x3
A = np.array([
    [2, 1, 3],
    [1, 0, 2],
    [4, 1, 8]
])
print("Matrice A :\n", A)
# 2. Calcul du déterminant
det_A = np.linalg.det(A)
print("\nDéterminant de A :", det_A)
# 3. Calcul de l'inverse (si possible)
if det_A != 0:
    inv_A = np.linalg.inv(A)
    print("\nInverse de A :\n", inv_A)
else:
    print("\nLa matrice n'est pas inversible (déterminant = 0)")
```

Exercice2

```python
import numpy as np
# Génération de 50 nombres aléatoires entre 0 et 100
donnees = np.random.randint(0, 101, 50)
print("Données :")
print(donnees)
# Calcul de la moyenne
moyenne = np.mean(donnees)
# Calcul de la médiane
mediane = np.median(donnees)
# Calcul de l'écart type
ecart_type = np.std(donnees)
# Affichage des résultats
print("\nMoyenne :", moyenne)
print("Médiane :", mediane)
print("Écart type :", ecart_type)
```

```python
import numpy as np
# Création d'un tableau de dates pour janvier 2023
dates = np.arange('2023-01-01', '2023-02-01', dtype='datetime64[D]')
print("Dates originales :")
print(dates)
# Conversion au format AAAA/MM/JJ
dates_formatees = [str(date).replace('-', '/') for date in dates]
print("\nDates au format AAAA/MM/JJ :")
for date in dates_formatees:
    print(date)
```

```python
import numpy as np
import pandas as pd
# Génération de 10 nombres aléatoires entre 1 et 100
donnees = np.random.randint(1, 101, 10)
# Création du DataFrame
df = pd.DataFrame({
    'Valeurs': donnees
})
print("DataFrame :")
print(df)
# Sélection conditionnelle : valeurs supérieures à 50
valeurs_sup_50 = df[df['Valeurs'] > 50]
print("\nValeurs supérieures à 50 :")
print(valeurs_sup_50)
# Fonctions d'agrégation
somme = df['Valeurs'].sum()
moyenne = df['Valeurs'].mean()
print("\nSomme :", somme)
print("Moyenne :", moyenne)
```

Exercice6

```python
import numpy as np
# Productivité avant la formation
productivity_before = np.random.normal(loc=50, scale=10, size=30)
# Productivité après la formation
productivity_after = productivity_before + np.random.normal(loc=5, scale=3, size=30)
# Moyennes
mean_before = np.mean(productivity_before)
mean_after = np.mean(productivity_after)
# Écart-type
std_before = np.std(productivity_before)
std_after = np.std(productivity_after)
# Différence moyenne
mean_difference = np.mean(productivity_after - productivity_before)
print("Moyenne avant :", mean_before)
print("Moyenne après :", mean_after)
print("Écart-type avant :", std_before)
print("Écart-type après :", std_after)
print("Amélioration moyenne :", mean_difference)
```

```python
import numpy as np
import pandas as pd
# Génération des dates de l'année 2023
dates = pd.date_range(start='2023-01-01', end='2023-12-31', freq='D')
# Génération de valeurs aléatoires associées aux dates
valeurs = np.random.randint(0, 100, len(dates))
# Création du DataFrame
df = pd.DataFrame({
    'Date': dates,
    'Valeur': valeurs
})
# Définir la colonne Date comme index
df.set_index('Date', inplace=True)
# Segmentation des données
jan_mars = df['2023-01':'2023-03']
avr_juin = df['2023-04':'2023-06']
juil_sep = df['2023-07':'2023-09']
oct_dec = df['2023-10':'2023-12']
# Affichage
print("Janvier à Mars")
print(jan_mars.head())
print("\nAvril à Juin")
print(avr_juin.head())
print("\nJuillet à Septembre")
print(juil_sep.head())
print("\nOctobre à Décembre")
print(oct_dec.head())
```

Exercice9

```python
import numpy as np
import pandas as pd
# Création d'un tableau NumPy
tableau_numpy = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])
print("Tableau NumPy :")
print(tableau_numpy)
# Conversion NumPy -> Pandas DataFrame
df = pd.DataFrame(
    tableau_numpy,
    columns=["A", "B", "C"]
)
print("\nDataFrame Pandas :")
print(df)
# Conversion Pandas DataFrame -> NumPy
tableau_converti = df.to_numpy()
print("\nTableau NumPy reconverti :")
print(tableau_converti)
```
