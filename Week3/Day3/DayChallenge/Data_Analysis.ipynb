3.Analyse statistique (NumPy + SciPy)

Mesures centrales

```python
df.mean(numeric_only=True)
df.median(numeric_only=True)
df.mode().iloc[0]
```

Variabilité

```python
df.var(numeric_only=True)
df.std(numeric_only=True)
df.max(numeric_only=True) - df.min(numeric_only=True)
```

Skewness & Kurtosis

```python
import scipy.stats as stats
stats.skew(df.select_dtypes(include=np.number))
stats.kurtosis(df.select_dtypes(include=np.number))
```

Test d’hypothèse (RAM vs prix)

```python
group0 = df[df['price_range'] == 0]['ram']
group3 = df[df['price_range'] == 3]['ram']
stats.ttest_ind(group0, group3)
```

Corrélation

```python
df.corr()['price_range'].sort_values(ascending=False)
```

4.Visualisation des données

Histogramme

```python
import matplotlib.pyplot as plt
df['battery_power'].hist()
plt.title("Battery Power Distribution")
plt.show()
```

Nuage de points

```python
plt.scatter(df['ram'], df['price_range'])
plt.title("RAM vs Price Range")
plt.xlabel("RAM")
plt.ylabel("Price Range")
plt.show()
```

Boxplot

```python
df.boxplot(column='ram', by='price_range')
plt.title("RAM by Price Range")
plt.show()
```

Heatmap corrélation

```python
import seaborn as sns
sns.heatmap(df.corr())
plt.title("Correlation Heatmap")
plt.show()
```

5.Synthèse des idées

The statistical tests and visualizations show that RAM is the most influential factor in mobile price classification.
Battery power and internal memory also play an important role in determining price categories, although their impact is less dominant than RAM.
Some features show weak or negligible correlation with the target variable, indicating limited influence on pricing.
The hypothesis tests confirm that differences between price groups are statistically significant.
An interesting finding is that not all technical specifications contribute equally to pricing, with RAM being the strongest discriminator among all features.
