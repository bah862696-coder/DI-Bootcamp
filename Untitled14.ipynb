# Partie 5 — Découpage des documents

Le découpage (splitting) est essentiel car les modèles de langage ont une fenêtre de contexte limitée. De plus, des segments plus courts permettent une recherche plus précise dans la base de données vectorielle.

* **chunk_size** : Le nombre maximum de caractères par morceau.
* **chunk_overlap** : Le nombre de caractères partagés entre deux morceaux consécutifs pour éviter de perdre le contexte à la coupure.

```python
def test_splitter(docs, size, overlap):
    splitter = RecursiveCharacterTextSplitter(chunk_size=size, chunk_overlap=overlap)
    split_docs = splitter.split_documents(docs)
    avg_len = sum(len(d.page_content) for d in split_docs) / len(split_docs)
    print(f"--- Configuration: Size={size}, Overlap={overlap} ---")
    print(f"Nombre de chunks : {len(split_docs)}")
    print(f"Longueur moyenne : {avg_len:.2f} caractères")
    print(f"Exemple de chunk : {split_docs[0].page_content[:100]}...")
    print("-" * 50)
    return split_docs

# Test des 3 configurations demandées
# Note: Nous utilisons un échantillon de 1000 documents pour accélérer le traitement du split si nécessaire
sample_docs = processed_docs[:1000]

docs_conf1 = test_splitter(sample_docs, 1000, 150)
docs_conf2 = test_splitter(sample_docs, 700, 100)
docs_conf3 = test_splitter(sample_docs, 500, 50)

# Nous choisissons la configuration 1 (1000/150) pour la suite comme suggéré dans l'énoncé original
final_docs = docs_conf1
```

# Partie 6 — Création des embeddings

Un **embedding** est une représentation vectorielle d'un texte. Des textes ayant un sens similaire auront des vecteurs proches dans l'espace mathématique. Nous utilisons le modèle `all-MiniLM-L6-v2` qui est léger et performant.

```python
model_path = "sentence-transformers/all-MiniLM-L6-v2"
model_kwargs = {'device': 'cpu'} # Changez en 'cuda' si une GPU est disponible
encode_kwargs = {'normalize_embeddings': False}

embeddings = HuggingFaceEmbeddings(
    model_name=model_path,
    model_kwargs=model_kwargs,
    encode_kwargs=encode_kwargs
)

# Test de l'embedding
test_text = "This is a test document."
query_result = embeddings.embed_query(test_text)

print(f"Dimension de l'embedding : {len(query_result)}")
print(f"3 premières valeurs : {query_result[:3]}")
```

# Partie 7 — Construction du Vector Store

Nous utilisons **FAISS** (Facebook AI Similarity Search), une bibliothèque qui permet une recherche de similarité efficace dans des ensembles de vecteurs denses. C'est ici que nos documents indexés (sous forme de vecteurs) seront stockés pour être retrouvés plus tard.

```python
# Création de l'index FAISS à partir des documents découpés et des embeddings
# Cette étape peut prendre quelques instants
vector_db = FAISS.from_documents(final_docs, embeddings)

print(f"Nombre de fragments de documents indexés dans FAISS : {len(final_docs)}")
print("Le Vector Store a été construit avec succès.")
```

# Partie 8 — Création du Retriever

Le **Retriever** est l'interface qui permet de récupérer les documents les plus pertinents pour une question donnée. Le paramètre **k** définit le nombre de documents à extraire.

```python
def test_retriever(k_val):
    retriever = vector_db.as_retriever(search_kwargs={"k": k_val})
    query = "What is cheesemaking?"
    results = retriever.get_relevant_documents(query)
    print(f"--- Test Retriever avec k={k_val} ---")
    print(f"Nombre de documents récupérés : {len(results)}")
    if results:
        print(f"Source du premier doc : {results[0].metadata.get('source')}")
    print("-" * 50)
    return retriever

# Comparaison des valeurs de k
retriever_k2 = test_retriever(2)
retriever_k4 = test_retriever(4)
retriever_k6 = test_retriever(6)

# Nous utiliserons k=4 par défaut pour un bon équilibre entre contexte et précision
final_retriever = retriever_k4
```

# Partie 9 — Vérification de la récupération

Avant de solliciter le LLM, il est crucial de vérifier que notre Retriever trouve bien les bons documents. Si les documents ne sont pas pertinents, la réponse du modèle ne le sera pas non plus (principe du *Garbage In, Garbage Out*).

```python
test_questions = [
    "What is cheesemaking?",
    "How to make bread?"
]

for q in test_questions:
    print(f"\n🔍 Question : {q}")
    docs = final_retriever.invoke(q)
    for idx, d in enumerate(docs):
        print(f"\n  Document {idx+1} (Source: {d.metadata['index']}):")
        print(f"  Contenu: {d.page_content[:200]}...")
        print(f"  Catégorie: {d.metadata['category']}")
```

# Partie 10 — Chargement du modèle Hugging Face

Nous utilisons `Intel/dynamic_tinybert`, un modèle extrêmement léger optimisé pour les tâches de Question-Answering, idéal pour un environnement CPU comme la version gratuite de Colab.

```python
model_name = "Intel/dynamic_tinybert"

# Chargement du tokenizer et du modèle
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForQuestionAnswering.from_pretrained(model_name)

# Création du pipeline Hugging Face
# Note : 'question-answering' est une tâche spécifique qui extrait la réponse du contexte
qa_pipeline = pipeline(
    "question-answering",
    model=model,
    tokenizer=tokenizer,
    device=-1 # -1 pour CPU, 0 pour GPU
)

# Wrapper LangChain pour intégrer le pipeline dans une chaîne
llm = HuggingFacePipeline(
    pipeline=qa_pipeline,
    model_kwargs={"temperature": 0, "max_length": 512}
)

print("Modèle LLM chargé et prêt.")
```

# Partie 11 — Construction de la chaîne RAG

Nous utilisons `RetrievalQA` avec le type de chaîne `refine` ou `stuff`. Ici, nous configurons la chaîne pour qu'elle utilise notre retriever et notre modèle.

```python
# Configuration de la chaîne de Question-Réponse
qa_chain = RetrievalQA.from_chain_type(
    llm=llm,
    chain_type="stuff", # 'stuff' envoie tous les docs au prompt, simple et efficace ici
    retriever=final_retriever,
    return_source_documents=True
)

print("Chaîne RAG (RetrievalQA) initialisée.")
```

# Partie 13 — Analyse finale

### Fonctionnement du système RAG
Le **RAG (Retrieval-Augmented Generation)** combine la puissance de recherche d'une base de données et la capacité de synthèse d'un modèle de langage. Au lieu de se fier uniquement aux connaissances internes du LLM (qui peuvent être datées ou erronées), on lui fournit des documents fiables en temps réel.

### Composants clés :
- **Embeddings** : Ils transforment le texte en vecteurs. C'est le 'pont' qui permet aux machines de comparer le sens sémantique des phrases.
- **FAISS** : C'est notre bibliothèque de recherche. Elle permet de trouver instantanément les vecteurs les plus proches parmi des milliers de documents.
- **Chunking (Découpage)** : Un `chunk_size` trop grand dilue l'information. Un `chunk_overlap` trop petit risque de couper une information importante en deux. L'équilibre (ex: 1000/150) est crucial.
- **Le paramètre K** : Il définit la quantité de contexte. Trop de documents peuvent saturer le modèle ou inclure du bruit.

### Réduction des hallucinations
Un LLM seul a tendance à inventer une réponse plausible s'il ne connaît pas le sujet. En RAG, le modèle est contraint par le contexte fourni. S'il ne trouve pas la réponse dans les documents, on peut le configurer pour qu'il dise 'Je ne sais pas', réduisant ainsi drastiquement les hallucinations.

### Limites et améliorations
- **Limites** : Le modèle `TinyBERT` est très rapide mais peut manquer de nuances par rapport à des modèles plus gros (Llama 3, Mistral).
- **Améliorations** : Utiliser un 'Reranker' pour affiner les résultats du retriever, ou tester des méthodes de découpage sémantique.

# Partie 3 — Chargement du dataset

Nous utilisons le jeu de données `databricks/databricks-dolly-15k`. Ce dataset contient des paires d'instructions et de réponses, souvent accompagnées d'un contexte. C'est ce contexte que nous utiliserons comme source d'information pour notre système RAG.

```python
dataset_name = "databricks/databricks-dolly-15k"
page_content_column = "context"

# Initialisation du chargeur
# HuggingFaceDatasetLoader charge les données directement depuis le Hub Hugging Face
loader = HuggingFaceDatasetLoader(dataset_name, page_content_column)
raw_data = loader.load()

print(f"Nombre total de documents chargés : {len(raw_data)}")
print(f"Exemple de colonnes disponibles dans le premier document : {list(raw_data[0].metadata.keys())}")

# Affichage de deux exemples pour vérifier la structure
print("\n--- Exemple 1 ---")
print(raw_data[0].page_content[:200] + "...")
print("\n--- Exemple 2 ---")
print(raw_data[1].page_content[:200] + "...")
```

# Partie 4 — Conversion et Enrichissement des Documents

Bien que le loader crée déjà des objets `Document`, nous allons nous assurer qu'ils sont correctement formatés avec des métadonnées utiles comme la source, la catégorie et l'index original.

```python
processed_docs = []

for i, doc in enumerate(raw_data):
    # On ne garde que les documents ayant un contenu non vide
    if doc.page_content.strip():
        metadata = {
            "source": dataset_name,
            "index": i,
            "category": doc.metadata.get("category", "N/A"),
            "instruction": doc.metadata.get("instruction", "N/A")
        }
        new_doc = Document(page_content=doc.page_content, metadata=metadata)
        processed_docs.append(new_doc)

print(f"Nombre de documents après filtrage : {len(processed_docs)}")
print("\nExemple de document traité :")
print(f"Contenu : {processed_docs[0].page_content[:150]}...")
print(f"Métadonnées : {processed_docs[0].metadata}")
```
