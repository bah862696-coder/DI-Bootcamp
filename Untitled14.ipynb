# Solution du Défi Pratique : Techniques Avancées de Prompt Engineering

Ce document présente une solution complète et structurée aux exercices de contrôle de production de l'IA, appliquée au contexte de la communication d'entreprise sur la cybersécurité.

---

## Étape 1 — Création d'un prompt avancé

**Objectif :** Transformer un texte technique et dense en un contenu de micro-apprentissage fluide, tout en respectant des contraintes strictes de format et de ton.

### Prompt Optimisé

```markdown
# RÔLE
Tu es un Spécialiste Senior en Communication Interne et Formation au sein d'une entreprise technologique globale. Ton expertise réside dans la vulgarisation de politiques complexes.

# CONTEXTE
Nous mettons à jour notre politique de sécurité informatique. Le texte source est trop formel pour notre newsletter interne.

# TEXTE SOURCE
"Les employés doivent s'assurer que tout accès distant aux systèmes internes est établi via le VPN sécurisé approuvé. En aucun cas, des connexions non sécurisées ou des appareils personnels dépourvus de protection des terminaux ne doivent être utilisés pour accéder à des données confidentielles ou à des communications sensibles."

# INSTRUCTIONS
Réécris le contenu ci-dessus pour un public d'employés non techniques.
1. Ton : Amical, clair et bienveillant.
2. Format : Utilise exclusivement une liste à puces.
3. Méthode : Paraphrase intégralement le texte. Ne cite aucune phrase telle quelle.

# CONTRAINTES
- Longueur : Maximum 75 mots au total.
- Fidélité : Conserve toutes les obligations de sécurité sans altérer le sens original.
- Exclusion : N'ajoute aucune salutation ou introduction, commence directement par les puces.
```

### Justification
Ce prompt surpasse une consigne simple car il utilise le **Role Prompting** pour définir le standard de qualité, et le **Constraint Prompting** pour éliminer les fioritures (comme les salutations) qui consommeraient inutilement le budget de mots. En séparant le texte source des instructions, on réduit la confusion cognitive du modèle.

## Étape 3 — Détection et réduction des hallucinations

**Objectif :** Empêcher l'IA d'inventer des outils (ex: Authentification MFA) ou des recommandations non présentes dans la source.

### Prompt avec Contraintes Strictes (Anti-Hallucination)

```markdown
# INSTRUCTION STRICTE
Agis en tant que rédacteur technique factuel. Ta mission est de paraphraser le texte source ci-dessous avec une tolérance zéro pour l'ajout d'informations.

# RÈGLES DE CONTRÔLE
- UTILISE UNIQUEMENT les faits présents dans le texte source.
- INTERDICTION d'ajouter des conseils de sécurité supplémentaires (ex: mots de passe, MFA).
- INTERDICTION d'inventer des noms de technologies ou de logiciels.
- Si une information n'est pas explicitement mentionnée, ne l'infère pas.
- Ton : Neutre et instructif.

# TEXTE SOURCE
[Insérer le texte de la politique]
```

### Analyse Technique
Ici, nous utilisons le **Negative Prompting** (Interdictions explicites) et l'**ancrage textuel strict**. En limitant l'espace de recherche du modèle à la source fournie, on réduit drastiquement la probabilité de "complétion créative", cause majeure des hallucinations.

## Étape 4 — Adaptation pour un public de stagiaires juniors

### Prompt pour Public Junior

```markdown
# CIBLE
Stagiaires juniors (nouvellement arrivés, profil non technique).

# INSTRUCTIONS DE RÉDACTION
Réécris la politique de sécurité avec un ton grand frère/grande sœur, encourageant et simple.
- Maximum 4 puces.
- Phrases courtes (sujet-verbe-complément).
- Zéro jargon (remplace VPN par "réseau sécurisé de l'entreprise", remplace terminal par "ordinateur ou téléphone").

# OBJECTIF
Faire en sorte que le stagiaire se sente responsable et guidé plutôt que surveillé.
```

### Pourquoi ce prompt fonctionne ?
Il définit une **personnalité** (ton encourageant) et impose une **limitation lexicale** (zéro jargon). Cela force le modèle à choisir des synonymes accessibles, facilitant l'onboarding des nouveaux arrivants.

## Étape 5 — Variante utilisant une citation

### Prompt d'extraction

```markdown
# TÂCHE
Extrais l'unique phrase du texte source qui définit l'obligation principale de sécurité.

# CONTRAINTES
- Citation directe exacte.
- Aucun ajout, aucune modification de ponctuation.
- Aucune introduction.
```

### Analyse de l'usage des citations

1.  **Quand préférer la citation ?** Dans les documents juridiques, les signatures de mails de conformité ou lors de rappels disciplinaires où l'ambiguïté doit être nulle.
2.  **Risques ?** Une citation brute peut être perçue comme agressive (ton autoritaire) et peut ne pas être comprise par des employés non techniques si elle contient trop de jargon.
3.  **Comparaison :**
    *   *Citation :* Incontestable, précise, mais rigide et parfois obscure.
    *   *Paraphrase :* Accessible, engageante, mais risque de légère dérive de sens.

# Analyse finale : Synthèse du Prompt Engineering

Le **Prompt Engineering** n'est pas simplement une demande, c'est une architecture de contrôle.

*   **Qualité :** Il améliore la production en guidant le modèle vers des structures de données (format) et de pensée (rôle) spécifiques.
*   **Contraintes :** Elles sont essentielles pour garantir que l'IA reste dans les limites de la conformité et de l'utilisabilité.
*   **Adaptation :** Un prompt mal ciblé crée une barrière à l'information. Adapter le ton au public cible assure l'adoption des politiques par les employés.
*   **Hallucinations :** La réduction des hallucinations passe par des instructions de "fermeture du domaine" (ancrage source).
*   **Choix stratégique :** On utilisera la paraphrase pour éduquer et la citation pour faire autorité.

**Ce flux de travail démontre que la maîtrise de l'IA ne réside pas dans la puissance du modèle, mais dans la précision de l'intention transmise.**
