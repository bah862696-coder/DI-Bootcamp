## 3. Préparer l'ensemble de données pour l'entraînement

```python
import torch
from torch.utils.data import Dataset, DataLoader

# Définir la taille de la séquence (regarder en arrière pour prédire l'avenir)
sequence_length = 60 # Par exemple, utiliser les 60 jours précédents pour prédire le jour suivant

# Séparer les caractéristiques (X) et la cible (y)
# X sera toutes les colonnes sauf 'Target'
# y sera la colonne 'Target'

X = df_scaled.drop(columns=['Target']).values
y = df_scaled['Target'].values.reshape(-1, 1) # Reshape pour qu'il ait 2 dimensions

# Fonction pour créer des séquences pour les modèles de séries temporelles
def create_sequences(data, target, seq_length):
    xs, ys = [], []
    for i in range(len(data) - seq_length):
        x = data[i:(i + seq_length)]
        y_val = target[i + seq_length]
        xs.append(x)
        ys.append(y_val)
    return np.array(xs), np.array(ys)

X_seq, y_seq = create_sequences(X, y, sequence_length)

print(f"Forme de X_seq après séquençage : {X_seq.shape}") # (Nombre d'échantillons, Longueur de séquence, Nombre de caractéristiques)
print(f"Forme de y_seq après séquençage : {y_seq.shape}") # (Nombre d'échantillons, 1)

# Diviser l'ensemble de données en ensembles d'entraînement, de validation et de test
# La division doit être chronologique pour les séries temporelles

train_size = int(len(X_seq) * 0.7)
val_size = int(len(X_seq) * 0.15)
test_size = len(X_seq) - train_size - val_size

X_train, y_train = X_seq[:train_size], y_seq[:train_size]
X_val, y_val = X_seq[train_size:train_size + val_size], y_seq[train_size:train_size + val_size]
X_test, y_test = X_seq[train_size + val_size:], y_seq[train_size + val_size:]

print(f"\nTrain set shapes: X={X_train.shape}, y={y_train.shape}")
print(f"Validation set shapes: X={X_val.shape}, y={y_val.shape}")
print(f"Test set shapes: X={X_test.shape}, y={y_test.shape}")
```

```python
# Créer une classe PyTorch personnalisée Dataset
class StockDataset(Dataset):
    def __init__(self, X, y):
        # Convertir les données NumPy en tenseurs PyTorch
        self.X = torch.from_numpy(X).float()
        self.y = torch.from_numpy(y).float()

    def __len__(self):
        return len(self.X)

    def __getitem__(self, idx):
        return self.X[idx], self.y[idx]

# Instancier les ensembles de données pour l'entraînement, la validation et les tests
train_dataset = StockDataset(X_train, y_train)
val_dataset = StockDataset(X_val, y_val)
test_dataset = StockDataset(X_test, y_test)

print(f"\nNombre d'échantillons dans l'ensemble d'entraînement: {len(train_dataset)}")
print(f"Nombre d'échantillons dans l'ensemble de validation: {len(val_dataset)}")
print(f"Nombre d'échantillons dans l'ensemble de test: {len(test_dataset)}")
```

```python
# Utiliser DataLoader pour créer des ensembles de données itérables
batch_size = 64 # Taille du lot pour l'entraînement

train_loader = DataLoader(train_dataset, batch_size=batch_size, shuffle=False) # Pas besoin de mélanger pour les séries temporelles
val_loader = DataLoader(val_dataset, batch_size=batch_size, shuffle=False)
test_loader = DataLoader(test_dataset, batch_size=batch_size, shuffle=False)

print("\nDataLoaders créés.")
# Vérifier un exemple de lot
for features, targets in train_loader:
    print(f"Forme des caractéristiques dans un lot d'entraînement: {features.shape}") # (batch_size, sequence_length, num_features)
    print(f"Forme des cibles dans un lot d'entraînement: {targets.shape}") # (batch_size, 1)
    break
```

## 4. Définir le modèle LSTM

```python
import torch.nn as nn

class LSTMModel(nn.Module):
    def __init__(self, input_size, hidden_size, num_layers, output_size, dropout_rate=0.5):
        super(LSTMModel, self).__init__()
        self.hidden_size = hidden_size
        self.num_layers = num_layers

        # La couche LSTM prend en entrée (batch_size, sequence_length, input_size)
        # et produit (batch_size, sequence_length, hidden_size) pour la sortie
        # et (num_layers, batch_size, hidden_size) pour l'état caché et cellulaire.
        self.lstm = nn.LSTM(input_size, hidden_size, num_layers, batch_first=True, dropout=dropout_rate)

        # Couche Dropout pour la régularisation
        self.dropout = nn.Dropout(dropout_rate)

        # La couche linéaire pour la prédiction finale
        self.fc = nn.Linear(hidden_size, output_size)

    def forward(self, x):
        # Initialiser l'état caché et l'état de la cellule avec des zéros
        h0 = torch.zeros(self.num_layers, x.size(0), self.hidden_size).to(x.device)
        c0 = torch.zeros(self.num_layers, x.size(0), self.hidden_size).to(x.device)

        # Propager l'entrée à travers le LSTM
        # out: (batch_size, sequence_length, hidden_size)
        # hn, cn: (num_layers, batch_size, hidden_size)
        out, _ = self.lstm(x, (h0, c0))

        # Récupérer la sortie de la dernière étape temporelle
        # out[:, -1, :]: (batch_size, hidden_size)
        out = self.dropout(out[:, -1, :])

        # Passer par la couche entièrement connectée
        out = self.fc(out)
        return out

# Définir les hyperparamètres du modèle
input_size = X_train.shape[2] # Nombre de caractéristiques
hidden_size = 50 # Nombre d'unités dans la couche cachée du LSTM
num_layers = 2 # Nombre de couches LSTM empilées
output_size = 1 # Prédiction d'une seule valeur (cours de clôture)
dropout_rate = 0.2 # Taux de dropout

# Instancier le modèle
model = LSTMModel(input_size, hidden_size, num_layers, output_size, dropout_rate)

# Afficher l'architecture du modèle
print("\nArchitecture du modèle LSTM:")
print(model)
print(f"Nombre total de paramètres du modèle: {sum(p.numel() for p in model.parameters() if p.requires_grad)}")
```

## 5. Entraîner le modèle

```python
import torch.optim as optim

# Définir l'appareil (GPU si disponible, sinon CPU)
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
model.to(device)

# Configurer l'optimiseur et la fonction de perte
criterion = nn.MSELoss() # Erreur quadratique moyenne pour la régression
optimizer = optim.Adam(model.parameters(), lr=0.001) # Optimiseur Adam avec un taux d'apprentissage de 0.001

# Paramètres d'entraînement
num_epochs = 20 # Nombre d'époques d'entraînement

# Listes pour stocker les pertes d'entraînement et de validation
train_losses = []
val_losses = []

print(f"Début de l'entraînement sur l'appareil: {device}")

# Boucle d'entraînement
for epoch in range(num_epochs):
    model.train() # Mettre le modèle en mode entraînement
    running_train_loss = 0.0
    for batch_idx, (data, target) in enumerate(train_loader):
        data, target = data.to(device), target.to(device)

        optimizer.zero_grad() # Remettre les gradients à zéro
        outputs = model(data) # Passer les données au modèle
        loss = criterion(outputs, target) # Calculer la perte
        loss.backward() # Rétropropagation
        optimizer.step() # Mettre à jour les poids
        running_train_loss += loss.item()

    avg_train_loss = running_train_loss / len(train_loader)
    train_losses.append(avg_train_loss)

    # Boucle de validation
    model.eval() # Mettre le modèle en mode évaluation
    running_val_loss = 0.0
    with torch.no_grad(): # Désactiver le calcul du gradient pour la validation
        for data, target in val_loader:
            data, target = data.to(device), target.to(device)
            outputs = model(data)
            loss = criterion(outputs, target)
            running_val_loss += loss.item()

    avg_val_loss = running_val_loss / len(val_loader)
    val_losses.append(avg_val_loss)

    print(f'Epoch [{epoch+1}/{num_epochs}], Train Loss: {avg_train_loss:.4f}, Val Loss: {avg_val_loss:.4f}')

print("\nEntraînement terminé.")

# Sauvegarder le modèle entraîné (optionnel)
torch.save(model.state_dict(), 'lstm_stock_predictor.pth')
print("Modèle sauvegardé sous 'lstm_stock_predictor.pth'")
```
