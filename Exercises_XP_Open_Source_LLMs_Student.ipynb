## Exercise 1: Open Source Levels Reflection
**Use:** Fully Open, Weights Released, Architecture Only; model components; fine-tuning/domain adaptation; healthcare compliance.

**Step-by-step**
1) Gather definitions for the three openness levels.  
2) Identify key characteristics (what is open; what you can/cannot do).  
3) Compare side by side in a table (what's open vs. impact).  
4) Write a 3?5 sentence comparative paragraph.  
5) Answer the healthcare prompt (1?2 sentences) about retraining on clinical data.

**Deliverables:** paragraph + healthcare answer.

```python
import pandas as pd

# Exercise 1: Definitions and key characteristics
open_source_levels = {
    "Fully Open": {
        "definition": "Models where the training code, data cleaning scripts, and weights are all public (e.g., Pythia, OLMo).",
        "what_is_open": ["Training data recipes", "Model weights and code"],
        "what_you_can_do": ["Full audit of data bias", "Reproduce training from scratch"],
        "what_you_cannot_do": ["Hide proprietary data if contributing back to some copyleft projects"],
    },
    "Weights Released": {
        "definition": "Pre-trained weights are available, but the specific training data or exact pipeline might be private (e.g., Mistral, Llama).",
        "what_is_open": ["Model weights", "Inference code"],
        "what_you_can_do": ["Fine-tune on private data", "Self-host locally"],
        "what_you_cannot_do": ["Know exactly which documents were in the training set"],
    },
    "Architecture Only": {
        "definition": "Only the paper or the structural design is public; weights must be purchased or accessed via API (common in research papers before release).",
        "what_is_open": ["Model layers/parameters count", "Attention mechanism details"],
        "what_you_can_do": ["Implement the structure in code", "Study the theoretical efficiency"],
        "what_you_cannot_do": ["Generate text without training it yourself first"],
    },
}
display(pd.DataFrame(open_source_levels).T)
```

```python
comparison_table = """| Openness level | What's open? | Impact on retraining/modifying |
| --- | --- | --- |
| Fully Open | Code, weights, and data info | Highest: Allows deep domain adaptation and bias auditing. |
| Weights Released | Model weights and config | High: Perfect for fine-tuning (SFT/LoRA) on private datasets. |
| Architecture Only | Design patterns | Low: Requires massive compute to actually use/train from zero. |"""
print(comparison_table)
```

```python
comparative_paragraph = """The three levels of openness represent a spectrum of transparency and control for developers. Fully open models provide the ultimate visibility for security-sensitive sectors, whereas weights-released models offer a pragmatic middle ground for commercial deployment. Architecture-only releases primarily benefit the research community by advancing structural innovation without necessarily providing a ready-to-use tool. Choosing between them depends on whether you need to audit the training data or simply apply the model to a new task."""
print(comparative_paragraph)
```

```python
healthcare_prompt_answer = """For retraining a model on sensitive clinical data, a 'Weights Released' or 'Fully Open' model is essential. This allows the model to be hosted on-premise to ensure HIPAA compliance while enabling fine-tuning to learn specific medical terminology."""
print(healthcare_prompt_answer)
```
