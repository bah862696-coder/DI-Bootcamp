## Exercise 2: Multi-Part Prompt for Quiz Generation
**Scenario**: Create an educational prompt for 7th graders about volcanic eruptions.

### Tasks
- Ask for: 2-bullet summary + 3 MCQs (1 correct, 2 distractors).
- Require friendly tone, simple vocab (11?13 y/o), and Google Slides-ready output.
- Explain 3 ways your prompt improves on: `Make a quiz for kids about this article.`

```python
quiz_prompt = """
Agis en tant qu'Enseignant de Sciences pour des élèves de 11 à 13 ans.

SOURCE : [Insérer le texte sur les éruptions volcaniques ici]

TACHES :
1. Résume l'article ci-dessus en exactement deux points clés simplifiés.
2. Génère 3 questions à choix multiples (QCM) basées sur l'article.

CONTRAINTES :
- Niveau : Vocabulaire simple, adapté aux enfants (5ème).
- Structure QCM : Pour chaque question, propose 1 bonne réponse et 2 distracteurs (mauvaises réponses).
- Ton : Amical et encourageant.
- Format de sortie : Présente le contenu sous forme de diapositives structurées (Titre, Contenu) pour une insertion facile dans Google Slides.
"""
print(quiz_prompt)
```

```python
improvements = [
    "Précision du public cible : En spécifiant l'âge (11-13 ans), le modèle ajuste le niveau de complexité du langage, évitant un contenu trop technique ou enfantin.",
    "Structure du format de sortie : Le prompt impose un format compatible avec Google Slides et une structure de QCM précise, ce qui élimine le besoin de remise en forme manuelle.",
    "Contrôle de la fidélité : En demandant un résumé en deux points et des distracteurs spécifiques, on réduit le risque d'informations superflues ou de questions mal structurées."
]
display(improvements)
```

## Exercise 6: Hallucination Spotting and Mitigation
**Scenario**: Model added an unsupported claim in a climate/marine article summary.

### Tasks
1) Prompt the model to avoid hallucinations using only provided text.
2) Rewrite with verification language.
3) List 2 mitigation strategies.
4) Name 2 domains where hallucinations are risky and why.

```python
no_hallucination_prompt = """
Résume l'article suivant sur la biodiversité marine.

REGLE D'OR : Utilise UNIQUEMENT les faits explicitement mentionnés dans le texte fourni.
Ne fais aucune supposition et n'ajoute aucune statistique externe.
"""
print(no_hallucination_prompt)
```

```python
verification_prompt = """
Analyse l'article joint.

INSTRUCTIONS :
- Pour chaque affirmation, cite le passage du texte source.
- Si une information demandée n'est pas présente, réponds : 'Information non disponible dans le document'.
- N'inclus pas la statistique sur la disparition de 50% des espèces si elle n'est pas écrite mot pour mot.
"""
print(verification_prompt)
```

```python
mitigation_strategies = [
    'Ancrage textuel (Grounding) : Forcer le modèle à ne citer que la source fournie.',
    'Instruction de réponse négative : Autoriser explicitement le modèle à dire qu\'il ne sait pas.'
]
risky_domains = {
    'médical': {'name': 'Santé', 'why': 'Une dose ou un diagnostic erroné peut mettre des vies en danger.'},
    'juridique': {'name': 'Droit', 'why': 'L\'invention de précédents judiciaires peut conduire à des erreurs de jugement graves.'},
}
display(mitigation_strategies, risky_domains)
```

# Conclusion et Synthèse

L'ingénierie des prompts (Prompt Engineering) est devenue une compétence cruciale pour exploiter pleinement le potentiel des LLM. Voici les points clés à retenir de ces exercices :

1.  **L'importance du Contexte** : Un modèle ne peut pas deviner vos intentions. Lui fournir un **rôle** (Expert, Enseignant, Analyste) et définir son **public** change radicalement la structure et le ton de la réponse.
2.  **La Précision des Instructions** : Les consignes vagues mènent à des réponses génériques. En décomposant une tâche en étapes claires et en utilisant des techniques comme le *Chain of Thought*, on améliore la cohérence logique.
3.  **Le Pouvoir des Contraintes** : Limiter la longueur, imposer un format (JSON, Markdown, Slides) ou un vocabulaire spécifique permet d'obtenir un résultat directement exploitable, sans retouche manuelle.
4.  **Maîtrise de la Fiabilité** : Pour limiter les **hallucinations**, l'utilisation de techniques d'*ancrage textuel* (grounding) et l'autorisation explicite de répondre "Je ne sais pas" sont essentielles, particulièrement dans des domaines critiques comme la santé ou le droit.

En suivant ces principes, vous transformez l'IA d'un simple gadget de conversation en un outil de productivité puissant et fiable.
