# Exercice 1 : Introduction à l’analyse des données

## Introduction
L’analyse des données est un processus qui consiste à collecter, organiser, nettoyer et interpréter des données afin d’obtenir des informations utiles. Avec l’évolution rapide des technologies numériques, les entreprises, les gouvernements et les organisations produisent une grande quantité de données chaque jour. L’analyse des données permet donc de transformer ces informations brutes en connaissances exploitables pour prendre de meilleures décisions.
Aujourd’hui, l’analyse des données est devenue essentielle dans plusieurs domaines comme la santé, la finance, le commerce, les réseaux informatiques et l’intelligence artificielle. Grâce à des outils modernes tels que Python, Pandas, Matplotlib et les technologies de Big Data, il est possible d’analyser rapidement des millions de données.
---
## Qu’est-ce que l’analyse des données ?
L’analyse des données est l’ensemble des méthodes utilisées pour examiner des données afin de découvrir des tendances, résoudre des problèmes et aider à la prise de décision.
Elle comprend plusieurs étapes importantes :
- La collecte des données
- Le nettoyage des données
- L’organisation et la structuration
- L’analyse statistique
- La visualisation des résultats
Par exemple, une entreprise peut analyser les ventes de ses produits pour comprendre les préférences des clients et améliorer ses stratégies commerciales.
---
## Pourquoi l’analyse des données est-elle importante dans les contextes modernes ?
### 1. Aide à la prise de décision
Les organisations utilisent les données pour prendre des décisions plus précises et réduire les risques. Les données permettent d’obtenir des informations fiables au lieu de se baser uniquement sur des suppositions.
### 2. Détection des problèmes et amélioration des performances
Dans les réseaux informatiques et la cybersécurité, l’analyse des données aide à détecter les anomalies, les pannes et les cyberattaques. Cela améliore la sécurité et la stabilité des systèmes.
### 3. Prévision des tendances
Grâce à l’intelligence artificielle et au machine learning, les entreprises peuvent prévoir les comportements futurs des utilisateurs. Par exemple, Netflix ou Spotify recommandent des contenus selon les habitudes des utilisateurs.
### 4. Gain de temps et automatisation
Les outils modernes permettent d’automatiser l’analyse de grandes quantités de données, ce qui améliore la productivité et réduit les erreurs humaines.
---
## Trois domaines d’application de l’analyse des données
### 1. Santé
Dans le domaine médical, l’analyse des données permet de suivre les patients, détecter certaines maladies et améliorer les traitements. Les hôpitaux utilisent les données médicales pour prendre des décisions rapides et efficaces.
### 2. Commerce et marketing
Les entreprises analysent les habitudes d’achat des consommateurs afin de proposer des publicités ciblées et améliorer l’expérience client. Amazon utilise par exemple les données pour recommander des produits personnalisés.
### 3. Réseaux informatiques et cybersécurité
L’analyse des données est utilisée pour surveiller le trafic réseau et détecter les activités suspectes. Les administrateurs réseau peuvent analyser les journaux système afin d’identifier des attaques ou des tentatives d’intrusion.
---
## Conclusion
L’analyse des données joue aujourd’hui un rôle fondamental dans le monde moderne. Elle permet de transformer des données brutes en informations utiles pour résoudre des problèmes, optimiser les performances et prendre des décisions intelligentes. Avec le développement du Big Data, du cloud computing et de l’intelligence artificielle, les compétences en analyse des données deviennent de plus en plus importantes dans plusieurs secteurs professionnels.

```python
import pandas as pd

# Charger le fichier CSV
df = pd.read_csv("TimeSpleep.csv")

# Afficher les 5 premières lignes
df.head()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10,5))
plt.plot(df["Year"], df["Avg hrs per day sleeping"], marker='o')

plt.title("Évolution du sommeil moyen par année")
plt.xlabel("Année")
plt.ylabel("Heures de sommeil")

plt.grid()
plt.show()
```

```python
import seaborn as sns

plt.figure(figsize=(8,5))
sns.barplot(data=df, x="Sex", y="Avg hrs per day sleeping")

plt.title("Comparaison du sommeil par sexe")
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.barplot(data=df, x="Age Group", y="Avg hrs per day sleeping")

plt.xticks(rotation=45)
plt.title("Sommeil moyen selon l'âge")
plt.show()
```

```python
plt.figure(figsize=(10,5))
plt.bar(df["Year"], df["Standard Error"])

plt.title("Erreur standard du sommeil par année")
plt.xlabel("Année")
plt.ylabel("Standard Error")

plt.show()
```

# 📊 Exercice 6 : Identification des types de données

## 🎯 Objectif
Identifier si chaque source de données est **structurée** ou **non structurée**.
---
## 1️⃣ Rapports financiers d'une entreprise stockés dans un fichier Excel  
➡️ **Données structurées**
✔️ Justification :  
Les données sont organisées en lignes et colonnes avec un format fixe (tableau Excel), ce qui permet une analyse facile.
---
## 2️⃣ Photographies téléchargées sur une plateforme de médias sociaux  
➡️ **Données non structurées**
✔️ Justification :  
Les images ne suivent pas une structure tabulaire. Elles sont constituées de pixels et ne peuvent pas être directement organisées en colonnes.
---
## 3️⃣ Collection d'articles de presse sur un site web  
➡️ **Données non structurées**
✔️ Justification :  
Le texte est libre et non formaté sous forme de tableau exploitable directement.
---
## 4️⃣ Données d'inventaire dans une base de données relationnelle  
➡️ **Données structurées**
✔️ Justification :  
Les données sont stockées dans des tables avec des colonnes définies (produit, quantité, prix), ce qui facilite les requêtes.
---
## 5️⃣ Enregistrements d'entretiens issus d'une étude de marché  
➡️ **Données non structurées**

✔️ Justification :  
Les enregistrements audio ou texte libre ne suivent pas une structure prédéfinie exploitable directement.
---
## ✅ Résumé
- **Données structurées** : Excel, bases de données relationnelles  
- **Données non structurées** : images, textes, enregistrements audio

# 📊 Exercice 7 : Transformation de données non structurées

## 🎯 Objectif
Proposer des méthodes pour transformer des données non structurées en **données structurées**, avec justification.

---

## 1️⃣ Articles de blog sur les expériences de voyage  
➡️ **Méthode de transformation : Web scraping + NLP (traitement du langage naturel)**

✔️ Étapes :
- Extraire le texte des articles (scraping)
- Nettoyer les données (suppression des balises HTML, stop words)
- Utiliser le NLP pour identifier :
  - destinations
  - dates
  - sentiments
- Structurer dans un tableau (CSV ou base de données)

✔️ Exemple de structure finale :
| Auteur | Destination | Date | Sentiment | Thème |

✔️ Justification :
Cela permet d’extraire des informations utiles à partir de textes libres pour analyse (tendances de voyage, avis).

---

## 2️⃣ Enregistrements audio des appels au service client  
➡️ **Méthode de transformation : Transcription vocale (Speech-to-Text) + classification**

✔️ Étapes :
- Convertir l’audio en texte avec un outil de reconnaissance vocale
- Identifier les mots-clés (problème, demande, satisfaction)
- Classer les appels par catégories

✔️ Exemple de structure finale :
| Client | Problème | Durée appel | Résolution | Sentiment |

✔️ Justification :
La transcription permet de rendre les données exploitables et analysables automatiquement.

---

## 3️⃣ Notes manuscrites de brainstorming  
➡️ **Méthode de transformation : OCR + structuration manuelle ou IA**

✔️ Étapes :
- Utiliser un OCR (reconnaissance de texte)
- Corriger les erreurs de lecture
- Regrouper les idées par thèmes
- Structurer en liste ou tableau

✔️ Exemple de structure finale :
| Idée | Thème | Priorité | Responsable |

✔️ Justification :
Les notes manuscrites doivent être converties en texte numérique pour être organisées et analysées.

---

## 4️⃣ Tutoriel vidéo de cuisine  
➡️ **Méthode de transformation : Transcription vidéo + extraction d’étapes**

✔️ Étapes :
- Extraire l’audio de la vidéo
- Transcrire en texte
- Identifier les étapes de la recette
- Structurer en format recette

✔️ Exemple de structure finale :
| Étape | Action | Ingrédients | Temps |

✔️ Justification :
Transformer la vidéo en étapes structurées permet de créer une recette exploitable et réutilisable.

---

## 🎯 Conclusion
Les données non structurées peuvent être transformées grâce à :
- OCR (texte manuscrit)
- Speech-to-text (audio)
- NLP (texte)
- Video transcription (vidéo)

➡️ Objectif : rendre les données exploitables pour l’analyse et la prise de décision.
