# Exercice 1 : Comprendre la visualisation des données

## Pourquoi la visualisation des données est importante ?

La visualisation des données est essentielle dans l’analyse de données car elle permet de transformer des informations complexes en représentations graphiques simples et compréhensibles.

Elle est importante parce qu’elle permet de :

Comprendre rapidement de grandes quantités de données
Identifier des tendances et des évolutions
Repérer des anomalies ou des valeurs inhabituelles
Comparer facilement plusieurs ensembles de données
Faciliter la prise de décision basée sur les données

## Utilité d’un graphique linéaire

Un graphique linéaire est utilisé pour représenter l’évolution d’une variable en fonction du temps ou d’une progression continue.

### Il permet de :

Suivre une tendance dans le temps (hausse ou baisse)
Visualiser des évolutions continues
Comparer des changements sur plusieurs périodes

### Exemples :

 évolution des ventes mensuelles d’une entreprise
 évolution de la température sur une année
 croissance d’une population sur plusieurs années

```python
import matplotlib.pyplot as plt
# Jours de la semaine
jours = ["Lundi", "Mardi", "Mercredi", "Jeudi", "Vendredi", "Samedi", "Dimanche"]
# Températures
temperatures = [72, 74, 76, 80, 82, 78, 75]
# Graphique linéaire
plt.plot(jours, temperatures)
# Titre et étiquettes
plt.title("Variation de la température sur une semaine")
plt.xlabel("Jour")
plt.ylabel("Température (°F)")
# Affichage
plt.show()
```

```python
import matplotlib.pyplot as plt
# Mois
mois = ["Jan", "Fév", "Mar", "Avr", "Mai"]
# Ventes
ventes = [5000, 5500, 6200, 7000, 7500]
# Graphique à barres
plt.bar(mois, ventes)
# Titre et axes
plt.title("Ventes mensuelles d'un magasin")
plt.xlabel("Mois")
plt.ylabel("Montant des ventes ($)")
# Afficher le graphique
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
# Charger le fichier
df = pd.read_csv(r"C:\Users\home\Desktop\DI-Bootcamp\Week2\Day4\Student Mental health.csv")
# Calcul des proportions
anxiety_gender = pd.crosstab(
    df["Choose your gender"],
    df["Do you have Anxiety?"],
    normalize="index"
) * 100
# Graphique
anxiety_gender.plot(
    kind="bar",
    figsize=(8,5)
)
plt.title("Proportion d'étudiants souffrant d'anxiété selon le sexe")
plt.xlabel("Sexe")
plt.ylabel("Pourcentage (%)")
plt.legend(title="Anxiété")
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
# Charger le fichier
df = pd.read_csv(r"C:\Users\home\Desktop\DI-Bootcamp\Week2\Day4\Student Mental health.csv")
# Convertir les réponses aux crises de panique en valeurs numériques
df["Panic_Numeric"] = df["Do you have Panic attack?"].map({
    "Yes": 1,
    "No": 0
})
# Supprimer les lignes contenant des valeurs manquantes
df = df.dropna(subset=["Age", "Panic_Numeric"])
# Créer le nuage de points
plt.figure(figsize=(8, 5))
sns.scatterplot(
    data=df,
    x="Age",
    y="Panic_Numeric",
    color="purple",
    s=80,
    alpha=0.7
)
# Personnalisation
plt.title("Relation entre l'âge et les crises de panique")
plt.xlabel("Âge")
plt.ylabel("Crise de panique (0 = Non, 1 = Oui)")
# Affichage
plt.show()
```
