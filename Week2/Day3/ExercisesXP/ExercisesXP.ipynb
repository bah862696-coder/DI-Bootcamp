Exercice4

```python
import numpy as np
from scipy import stats
# données
data1 = np.random.normal(50, 10, 100)
data2 = np.random.normal(60, 10, 100)
# test t
t_stat, p_value = stats.ttest_ind(data1, data2)
print("T-statistique :", t_stat)
print("P-value :", p_value)
```

Ecxercice5

```python
import numpy as np
from scipy.stats import linregress
# données
house_sizes = np.array([50, 70, 80, 100, 120])
house_prices = np.array([150000, 200000, 210000, 250000, 280000])
# régression linéaire
slope, intercept, r, p, std_err = linregress(house_sizes, house_prices)
print("Pente :", slope)
print("Ordonnée à l'origine :", intercept)
# prédiction pour 90 m²
price_90 = slope * 90 + intercept
print("Prix estimé pour 90 m² :", price_90)
```

Exercice6

```python
from scipy import stats
# données
fertilizer_1 = [5, 6, 7, 6, 5]
fertilizer_2 = [7, 8, 7, 9, 8]
fertilizer_3 = [4, 5, 4, 3, 4]
# test ANOVA
f_stat, p_value = stats.f_oneway(fertilizer_1, fertilizer_2, fertilizer_3)
print("F-statistique :", f_stat)
print("P-value :", p_value)
```

```python
from scipy.stats import binom
# paramètres
n = 10      # nombre de lancers
p = 0.5     # probabilité d'avoir pile (ou face)
k = 5       # nombre de succès
# probabilité
resultat = binom.pmf(k, n, p)
print("Probabilité :", resultat)
```

Exercice8

```python
import pandas as pd
from scipy.stats import pearsonr, spearmanr
# données
data = pd.DataFrame({
    'age': [23, 25, 30, 35, 40],
    'income': [35000, 40000, 50000, 60000, 70000]
})
# Pearson (corrélation linéaire)
pearson_corr, _ = pearsonr(data["age"], data["income"])
# Spearman (corrélation par rang)
spearman_corr, _ = spearmanr(data["age"], data["income"])
print("Pearson :", pearson_corr)
print("Spearman :", spearman_corr)
```
