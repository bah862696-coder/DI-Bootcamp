```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler
from sklearn.decomposition import PCA

# =========================
# 1. Charger le dataset
# =========================

df = pd.read_csv("datascience_salaries.csv")

print(df.head())

# =========================
# 2. Normalisation Min-Max (salaire)
# =========================

scaler = MinMaxScaler()

df["salary_normalized"] = scaler.fit_transform(df[["salary"]])

print(df[["salary", "salary_normalized"]].head())

# =========================
# 3. Réduction de dimension (PCA)
# =========================

# garder seulement les colonnes numériques
numeric_df = df.select_dtypes(include=["int64", "float64"])

pca = PCA(n_components=2)

df_pca = pca.fit_transform(numeric_df)

print(df_pca[:5])

# =========================
# 4. Groupement par expérience
# =========================

result = df.groupby("experience_level")["salary"].agg(["mean", "median"])

print(result)
```
