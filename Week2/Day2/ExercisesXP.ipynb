```python
import pandas as pd
from sklearn.impute import SimpleImputer

# Charger le dataset Titanic
df = pd.read_csv("train.csv")

# 📌 Afficher les valeurs manquantes par colonne
print("Valeurs manquantes par colonne :")
print(df.isnull().sum())

# =====================================================
# 1️⃣ Suppression des lignes avec valeurs manquantes
# =====================================================

# Exemple : suppression des lignes où 'Embarked' est manquant
df_drop = df.dropna(subset=['Embarked'])

print("\nNombre de lignes après suppression :", df_drop.shape[0])

# =====================================================
# 2️⃣ Remplissage par une valeur constante
# =====================================================

# Exemple : remplacer les valeurs manquantes de Cabin
df['Cabin'] = df['Cabin'].fillna("Inconnu")

print("\nValeurs manquantes dans Cabin :")
print(df['Cabin'].isnull().sum())

# =====================================================
# 3️⃣ Imputation par la moyenne
# =====================================================

# Remplacer les valeurs manquantes de Age par la moyenne
age_imputer = SimpleImputer(strategy='mean')

df['Age'] = age_imputer.fit_transform(df[['Age']])

print("\nValeurs manquantes dans Age :")
print(df['Age'].isnull().sum())

# =====================================================
# 4️⃣ Imputation par la valeur la plus fréquente
# =====================================================

# Remplacer les valeurs manquantes de Embarked
embarked_imputer = SimpleImputer(strategy='most_frequent')

df['Embarked'] = embarked_imputer.fit_transform(df[['Embarked']]).ravel()

print("\nValeurs manquantes dans Embarked :")
print(df['Embarked'].isnull().sum())

# =====================================================
# Vérification finale
# =====================================================

print("\nValeurs manquantes restantes :")
print(df.isnull().sum())
```

```python
import pandas as pd
df = pd.read_csv("train.csv")
# =====================================================
# 1️⃣ Création de la variable FamilySize
# =====================================================
# FamilySize = nombre de frères/sœurs + parents/enfants + le passager lui-même
df["FamilySize"] = df["SibSp"] + df["Parch"] + 1

print(df[["SibSp", "Parch", "FamilySize"]].head())

# =====================================================
# 2️⃣ Extraction du titre depuis la colonne Name
# =====================================================

# Extraire le titre (Mr, Mrs, Miss, etc.)
df["Title"] = df["Name"].str.extract(r",\s*([^\.]+)\.")

print(df[["Name", "Title"]].head())

# =====================================================
# 3️⃣ Encodage des variables catégorielles
# =====================================================

# ----- One-Hot Encoding -----

# Encoder la colonne Embarked
df = pd.get_dummies(df, columns=["Embarked"], drop_first=True)

# Encoder la colonne Sex
df = pd.get_dummies(df, columns=["Sex"], drop_first=True)

# Encoder la colonne Title
df = pd.get_dummies(df, columns=["Title"], drop_first=True)

# =====================================================
# 4️⃣ Afficher le résultat
# =====================================================

print(df.head())

# =====================================================
# 5️⃣ Vérifier les nouvelles colonnes
# =====================================================

print(df.columns)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Charger le dataset
df = pd.read_csv("train.csv")

# ==========================
# 1️⃣ Visualiser les valeurs aberrantes
# ==========================

# Boxplot de Fare
plt.boxplot(df["Fare"].dropna())
plt.title("Valeurs aberrantes de Fare")
plt.show()

# ==========================
# 2️⃣ Détection avec IQR
# ==========================

Q1 = df["Fare"].quantile(0.25)
Q3 = df["Fare"].quantile(0.75)

IQR = Q3 - Q1

limite_basse = Q1 - 1.5 * IQR
limite_haute = Q3 + 1.5 * IQR

print("Limite basse :", limite_basse)
print("Limite haute :", limite_haute)

# Afficher les valeurs aberrantes
outliers = df[
    (df["Fare"] < limite_basse) |
    (df["Fare"] > limite_haute)
]

print("Nombre de valeurs aberrantes :", outliers.shape[0])

# ==========================
# 3️⃣ Traitement des valeurs aberrantes
# ==========================

# Supprimer les valeurs aberrantes
df_clean = df[
    (df["Fare"] >= limite_basse) &
    (df["Fare"] <= limite_haute)
]

# ==========================
# 4️⃣ Comparaison avant/après
# ==========================

print("Taille avant :", df.shape[0])
print("Taille après :", df_clean.shape[0])

# Nouveau boxplot
plt.boxplot(df_clean["Fare"].dropna())
plt.title("Après suppression des valeurs aberrantes")
plt.show()
```
