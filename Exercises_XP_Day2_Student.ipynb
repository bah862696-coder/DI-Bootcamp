- The transition to representation learning allowed models to capture deep semantic nuances without manual feature engineering, drastically reducing the human labor needed for new tasks. ## 🌟 Exercise 2: LLM Architecture and Application Scenarios
For each, describe (a) core architectural differences, (b) a real application, (c) why it fits.

### BERT
- **Architecture:** Bidirectional encoder-only; uses Masked Language Modeling (MLM) to look at both left and right context simultaneously.
- **Application:** Search Engine Query Understanding (e.g., Google Search).
- **Why:** It excels at understanding the specific intent and context of words in a sentence, which is crucial for retrieving relevant search results.

### GPT
- **Architecture:** Decoder-only; uses autoregressive causal language modeling, predicting the next token based strictly on previous tokens.
- **Application:** Creative Writing Assistants and Chatbots (e.g., ChatGPT).
- **Why:** Its causal nature makes it highly optimized for generating coherent, fluent, and creative sequences of text.

### T5
- **Architecture:** Encoder-decoder; treats every NLP task as a 'text-to-text' problem (input text -> output text).
- **Application:** Abstractive Document Summarization.
- **Why:** The encoder processes the full source document while the decoder generates a concise summary, making it perfect for transformation tasks.

### (Optional) Quick Generation Demo Across Temperatures
Note: Requires `transformers` and model download; skip if offline.


```python
from transformers import AutoTokenizer, AutoModelForCausalLM, pipeline
import torch

model_name = "gpt2"  # TODO: choose a causal LM
prompt = "Artificial intelligence will"  # TODO: write your own prompt

pipe = pipeline(
    "text-generation",
    model=AutoModelForCausalLM.from_pretrained(model_name),
    tokenizer=AutoTokenizer.from_pretrained(model_name),
    device=0 if torch.cuda.is_available() else -1,
)

for T in [0.2, 1.0, 1.5]:
    out = pipe(prompt, max_new_tokens=40, temperature=T, do_sample=True)
    print("--- Temperature:", T)
    print(out[0]["generated_text"])  # Inspect how style changes with T
```
