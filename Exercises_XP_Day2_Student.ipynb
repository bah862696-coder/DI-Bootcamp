## 🌟 Exercise 2: LLM Architecture and Application Scenarios
For each, describe (a) core architectural differences, (b) a real application, (c) why it fits.

### BERT
- BERT's core architecture is an **encoder-only Transformer** that processes input text bidirectionally. Unlike traditional LMs, it considers the entire context (both left and right) of a word to learn its representation. Its primary pre-training objectives are Masked Language Modeling (MLM), where random tokens are masked and the model predicts them, and Next Sentence Prediction (NSP) (historically used, now often omitted), where the model predicts if two sentences follow each other. This bidirectional context understanding is crucial for tasks requiring deep comprehension.
- Application: **Question Answering (Q&A)** on a given document.
- Why: BERT's bidirectional nature allows it to effectively understand the context of both the question and the document simultaneously. This deep contextual understanding enables it to pinpoint precise answer spans within the text, making it highly effective for extractive Q&A systems where the answer exists directly in the document.

### GPT
- GPT (Generative Pre-trained Transformer) is a **decoder-only Transformer** architecture. It is characterized by its autoregressive and causal language modeling approach, meaning it predicts the next token in a sequence based only on the preceding tokens. It does not have access to future tokens during training, making it highly suitable for generative tasks. Its training objective is typically to predict the next word in a sequence.
- Application: **Creative content generation**, such as writing stories, poems, or marketing copy.
- Why: GPT's causal, autoregressive nature makes it inherently excellent at generating coherent, contextually relevant, and creative text sequences from a given prompt. It excels at continuing a narrative or conversation naturally, making it ideal for applications where generating new, fluent text is the primary goal.

### T5
- T5 (Text-to-Text Transfer Transformer) utilizes an **encoder-decoder Transformer** architecture, treating every NLP problem as a "text-to-text" problem. This unified framework means that tasks like translation, summarization, or question answering are all framed as taking text as input and producing text as output, often preceded by a task-specific prefix (e.g., "translate English to German:"). This architecture allows for transfer learning across various tasks.
- Application: **Abstractive Summarization** of long articles or documents.
- Why: T5's encoder-decoder structure is perfectly suited for sequence-to-sequence tasks like summarization. The encoder can comprehend the full context of the input document, while the decoder can generate a coherent and concise summary that might not directly copy sentences from the original text (abstractive). The text-to-text framework allows for easy adaptation to this specific task by simply prepending "summarize:" to the input.


## 🌟 Exercise 3: Benefits and Ethics of Pre-training
- Benefits (explain each in your words):
  1) Improved generalization: Pre-training on vast, diverse datasets exposes models to a wide range of language patterns, grammar, and factual knowledge. This broad exposure helps them learn robust representations that generalize well to unseen data and new tasks, even those they weren't explicitly trained on.
  2) Less labeled data: Pre-trained models capture a significant amount of linguistic knowledge. When fine-tuning for a specific downstream task, they require significantly less task-specific labeled data compared to training a model from scratch, as they already have a strong foundation.
  3) Faster fine-tuning: Because pre-trained models have already learned general language understanding, fine-tuning them for a specific task is much faster than training from scratch. The fine-tuning process primarily involves adapting the pre-learned knowledge to the nuances of the new task, rather than learning everything from zero.
  4) Transfer learning: This is the core concept where knowledge gained from solving one problem (pre-training on a large corpus) is applied to a different but related problem (downstream NLP tasks). Pre-trained models act as powerful feature extractors or initializers, allowing for effective knowledge transfer and performance gains on new tasks.
  5) Robustness: The extensive pre-training process makes these models more resilient to noise, variations, and ambiguities in real-world data. They develop a deeper understanding of language, which contributes to more stable and reliable performance across diverse inputs and challenging scenarios.

- Ethical concerns: Pre-training LLMs on massive datasets often involves scraping data from the internet, which can contain a myriad of ethical issues. These include **bias** (reinforcing and amplifying societal prejudices present in the training data), **misinformation** and **disinformation** (generating plausible but false content), **misuse** (for spam, propaganda, deepfakes, or malicious code generation), and **privacy** (memorizing and reproducing sensitive personal information present in the training data).
- Mitigations: Addressing these concerns requires a multi-faceted approach. **Data curation** involves carefully selecting and filtering training data to reduce bias and harmful content. **Differential privacy** techniques can be applied during training to limit the memorization of specific data points. Implementing **safety filters** and content moderation mechanisms on model outputs can prevent the generation of harmful content. **Reinforcement Learning from Human Feedback (RLHF)** helps align model behavior with human values and safety guidelines. Regular **audits** and evaluations by independent bodies are crucial to identify and address emerging ethical risks.


## 🌟 Exercise 4: Transformer Architecture Deep Dive
### Self-Attention & Multi-Head Attention
- In **Self-Attention**, for each token in an input sequence, three vectors are created: a Query (Q), a Key (K), and a Value (V). These are generated by multiplying the token's embedding by three different weight matrices. To determine the importance of other tokens to the current token (Query), the Query vector is dot-product with all Key vectors in the sequence. The resulting scores are then scaled and passed through a **softmax** function, which converts them into probability distributions (weights) indicating how much attention each token should pay to every other token. Finally, these **softmax weights** are multiplied by the corresponding Value vectors, and the weighted Value vectors are summed up to produce the output for the current token. This process effectively allows each token to 'mix' information from all other tokens, prioritizing the most relevant ones.
- **Multiple heads** help by allowing the model to capture different aspects of relationships within the sequence simultaneously. Each head learns a distinct set of Q, K, and V weight matrices, enabling it to project the input into a different 'representation subspace'. This means one head might focus on syntactic relationships (e.g., subject-verb agreement), while another focuses on semantic dependencies (e.g., coreference resolution), and yet another on contextual nuances. By processing these diverse relationships in parallel and then concatenating and linearly transforming their outputs, multi-head attention provides a richer, more comprehensive understanding of the input sequence than a single head could.
- Example sentence (different from lesson):
  - Sentence: "The quick brown fox jumped over the lazy dog as it chased the rabbit."
  - Head 1 (Subject-Verb Agreement): A head might capture the relationship between "fox" and "jumped," understanding that "fox" is the subject performing the action. It would assign high attention scores from "jumped" to "fox."
  - Head 2 (Coreference Resolution): Another head might link "it" back to "fox," understanding that "it" refers to the fox, not the dog or the rabbit. It would assign a high attention score from "it" to "fox."
  - Head 3 (Semantic Relation - Agent/Patient): A third head could identify the object of the action, linking "jumped over" with "dog" (the entity being jumped over) and "chased" with "rabbit" (the entity being chased).

### Pre-training Objectives
- **MLM (Masked Language Modeling)** involves masking a percentage of tokens in the input sequence and training the model to predict the original masked tokens based on their surrounding context (both left and right). This objective is typically used in encoder-only architectures like BERT and focuses on learning deep bidirectional representations. **CLM (Causal Language Modeling)**, on the other hand, involves predicting the next token in a sequence given only the preceding tokens. It is an autoregressive objective, primarily used in decoder-only architectures like GPT, where the model generates text sequentially.
- **When to prefer MLM vs CLM:** Prefer **MLM** for tasks that require a deep, bidirectional understanding of the entire input sequence, such as sentiment analysis, named entity recognition, or question answering, where knowing the full context is crucial for accurate predictions. Prefer **CLM** for generative tasks where the goal is to produce new, coherent text sequentially, such as text generation, summarization, or translation, as it naturally learns to model conditional probabilities of token sequences.
- **NSP (Next Sentence Prediction):** Early BERT models used NSP as a secondary pre-training objective alongside MLM. It involved training the model to predict whether a given sentence B immediately followed sentence A in the original document. This was intended to help the model learn sentence-level relationships and discourse coherence, which was thought to be beneficial for tasks like Q&A and natural language inference. Many modern models avoid it because research (e.g., RoBERTa) found that NSP's contribution to overall performance was often minimal or even detrimental, and simpler objectives (like just MLM with longer sequences) proved more effective. The task itself was sometimes too easy or didn't sufficiently align with the needs of diverse downstream tasks, making its computational cost outweigh its benefits.

### Transformer Model Selection
- Sentiment on reviews: **Encoder-only** (e.g., BERT, RoBERTa) + justification: Sentiment analysis is a classification task that requires a deep, bidirectional understanding of the entire input review to grasp the nuances of sentiment. An encoder-only model excels at creating rich contextual embeddings for each word based on its full context, which is ideal for classifying the overall tone or emotion of a text.
- Conversational chatbot (creative responses): **Decoder-only** (e.g., GPT) + justification: Generating creative and coherent conversational responses is an autoregressive task. A decoder-only model is designed to predict the next token sequentially, conditioned on all preceding tokens, making it perfect for generating free-form, creative, and contextually appropriate text in a dialogue setting.
- Technical document translation (EN→ES): **Encoder-Decoder** (e.g., T5, BART) + justification: Machine translation is a classic sequence-to-sequence task. An encoder-decoder architecture allows the encoder to fully comprehend the source language document, building a rich representation, while the decoder can then generate the translation in the target language, leveraging the encoder's understanding. This separation of concerns (understanding vs. generation) is highly effective for translation.

### Positional Encoding
- Purpose: **Positional Encoding** is crucial in Transformer architectures because self-attention mechanisms inherently lack any notion of word order or position. Without it, the model would treat all words in a sequence as if they were in a bag-of-words, losing vital information about syntax, grammar, and sequential dependencies. Positional encodings, typically added to the word embeddings, inject information about the absolute or relative position of each token in the sequence, allowing the model to differentiate between identical words appearing at different positions and understand the sequence's structure.
- Example failure without positions: Consider the sentences "The dog bit the man" and "The man bit the dog." Without positional encoding, a Transformer model would produce identical representations for these two sentences (assuming the same word embeddings for "dog," "bit," and "man") because it would only see the presence of the words, not their order. Consequently, it would fail to capture the crucial difference in meaning (who is the subject and who is the object), leading to incorrect interpretations for tasks like sentiment analysis or question answering where causality or agent/patient roles are critical.

  - Justification: ELECTRA uses a more efficient pre-training objective called "replaced token detection" instead of masking. A generator replaces some tokens with plausible alternatives, and a discriminator is trained to identify which tokens have been replaced. This objective is much more computationally efficient during pre-training, allowing for smaller models to achieve performance comparable to much larger masked language models.
- Scenario 5 (efficient NLP, constrained environments): **ALBERT**
  - Justification: ALBERT (A Lite BERT) focuses on reducing memory consumption and speeding up training through two main parameter reduction techniques: factorized embedding parameterization and cross-layer parameter sharing. This makes it highly efficient in terms of memory and parameters, beneficial for constrained environments while still maintaining strong performance.

## 🌟 Exercise 6: Softmax Temperature: The Randomness Regulator
### 1) Temperature Scenarios
- **T=0.2:** At a very low temperature (e.g., 0.2), the softmax distribution becomes sharply peaked. The model will be highly confident in its top-ranked token and assign very low probabilities to alternative tokens. This results in highly deterministic, conservative, and less creative outputs. The text generated will likely be very coherent, predictable, and stick closely to the most probable sequences, potentially repeating phrases or falling into loops.
- **T=1.5:** At a high temperature (e.g., 1.5), the softmax distribution becomes flatter, making the probabilities of various tokens more uniform. This increases the likelihood of sampling less probable, more 'surprising' tokens. The model's outputs will be significantly more diverse, creative, and potentially less coherent or nonsensical at times. It encourages exploration of the model's vocabulary and can lead to highly imaginative but sometimes off-topic or grammatically incorrect text.
- **T=1.0:** A temperature of 1.0 (the default) means no modification is applied to the raw logits before the softmax. This maintains the original probability distribution learned by the model during training. The output will reflect the model's 'natural' balance between coherence and creativity, providing a generally fluent and contextually relevant text while still allowing for some variability.

### 2) Application Design
- **Bedtime stories (creativity vs coherence):** For bedtime stories, a balance between creativity and coherence is desired. I would start with a **moderate temperature, perhaps slightly above 1.0 (e.g., 1.0 to 1.2)**. This allows for imaginative elements and unexpected twists to keep children engaged (creativity) while ensuring the story remains logical, grammatically correct, and follows a narrative arc (coherence). If the story becomes too repetitive, I might incrementally increase the temperature; if it becomes too chaotic, I'd lower it slightly. I might also employ techniques like top-k or top-p sampling in conjunction with temperature to further fine-tune the sampling diversity.
- **Financial report summaries (accuracy/reliability):** For financial report summaries, **accuracy and reliability are paramount**, and creativity is undesirable. Therefore, I would set the **temperature very low, close to 0.0 (e.g., 0.1 to 0.3)**. This makes the model's output highly deterministic, forcing it to choose the most probable, factual tokens. The goal is to generate concise, precise, and objective summaries that directly reflect the information in the financial report without any fabrication or speculative language. In addition to low temperature, I might consider greedy decoding (T=0) or beam search to ensure the most probable and coherent sequence is chosen.

### 3) Temperature & Bias
- **How does temperature influence bias surfacing or dampening? Give a realistic example.**
  Temperature can significantly influence the surfacing or dampening of biases embedded within a language model. At **low temperatures (e.g., T=0.2)**, the model tends to choose the most probable tokens, which often correspond to the dominant patterns and stereotypes present in its training data. This can cause biases to surface more prominently because the model adheres strictly to the most common associations it has learned. Conversely, at **high temperatures (e.g., T=1.5)**, the model explores a wider range of less probable tokens. While this can introduce more creativity, it might also lead to sampling tokens that are further away from the common biased associations, potentially **dampening** obvious stereotypical outputs or, conversely, introducing new, less predictable forms of bias. It can also, however, sometimes surface more extreme or negative forms of bias if those exist in the tails of the probability distribution.

  **Realistic Example:** Imagine a model trained on a dataset where doctors are predominantly male and nurses are predominantly female.
  - If you prompt the model with "The doctor walked into the room. He told the nurse..." and use a **low temperature (T=0.2)**, the model is highly likely to complete the sentence with a pronoun reflecting the learned bias, e.g., "...she took his instructions." Here, the bias linking "nurse" to female is strongly reinforced because "she" is the most probable token given the training data.
  - If you use a **high temperature (T=1.5)** with the same prompt, the model might produce more varied outputs. While "she" might still be probable, there's a higher chance of sampling "he," "they," or even a name, thereby **dampening** the explicit gendered assumption associated with "nurse" in this specific instance, simply due to the increased randomness in token selection. However, this doesn't remove the underlying bias; it merely makes its manifestation less predictable and potentially less frequent for a single generation. The bias is still present in the model's learned probabilities, but high temperature allows for deviations from the most probable (and often most biased) path.


### (Optional) Quick Generation Demo Across Temperatures
Note: Requires `transformers` and model download; skip if offline.


```python
from transformers import AutoTokenizer, AutoModelForCausalLM, pipeline
import torch

model_name = "gpt2"
prompt = "Artificial intelligence will"

pipe = pipeline(
    "text-generation",
    model=AutoModelForCausalLM.from_pretrained(model_name),
    tokenizer=AutoTokenizer.from_pretrained(model_name),
    device=0 if torch.cuda.is_available() else -1,
)

for T in [0.2, 1.0, 1.5]:
    out = pipe(prompt, max_new_tokens=40, temperature=T, do_sample=True)
    print("--- Temperature:", T)
    print(out[0]["generated_text"])
```
